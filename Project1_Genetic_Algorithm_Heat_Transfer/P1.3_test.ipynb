```python
'''>>>>> start CodeP1.2F25 (P1.3-modified: 5-constant model with γ and P)
    V.P. Carey ME249, Fall 2025'''

from random import random
import math, numpy
from copy import deepcopy
import matplotlib.pyplot as plt
import numpy as np
# ---------- P1.3 additions ----------
G_EN = 9.8  # earth-normal gravity (m/s^2) used in g + n4*G_EN*gamma
# -----------------------------------

'''INITIALIZING PARAMETERS'''
n = []
ntemp = []
gen=[0]
n1avg = [0.0]
n2avg = [0.0]
n3avg = [0.0]
n4avg = [0.0]
n5avg = [0.0]
meanAFerr=[0.0]
aFerrmeanavgn=[0.0] 

#set program parameters
NGEN = 6000      #number of generations (steps)
MFRAC = 0.5      # fraction of median threshold

# Build storage arrays
for k in range(NGEN-1):
    gen.append(k+1)
    meanAFerr.append(0.0)
    aFerrmeanavgn.append(0.0)
    n1avg.append(0.0)
    n2avg.append(0.0)
    n3avg.append(0.0)
    n4avg.append(0.0)
    n5avg.append(0.0)

# ---------- P1.3 initial guesses (per handout) ----------
n0i =  -1.0  
n1i = 0.000476 
n2i = 3.028 
n3i = 0.2249 
n4i = 1.054  
n5i = 0.217
# --------------------------------------------------------

# Create initial population (NS rows) with small random perturbations
# NOTE: n[i][0] is fixed at -1 to carry the sign used in Ferr; others evolve
n =  [[-1., n1i+0.001*random(), n2i+0.1*random(), n3i+0.0001*random(), n4i+0.0001*random(), n5i+0.0001*random()]]
for i in range(NS):  # NS == ND in this program
    n.append([-1., n1i+0.0001*random(), n2i+0.001*random(), n3i+0.0001*random(), n4i+0.0001*random(), n5i+0.0001*random()])

ntemp = deepcopy(n)

# Error arrays
Ferr =  [[0.0]]  
Ferravgn =  [[0.0]]
aFerr =  [[0.0]]
aFerravgn =  [[0.0]]
for i in range(ND-1):
    Ferr.append([0.0]); aFerr.append([0.0])
    Ferravgn.append([0.0]); aFerravgn.append([0.0])

aFerrmeanavgnMin=1e30

# Trackers
n1min=n2min=n3min=n4min=n5min=0.0
aFerrta=0.0
n1ta=n2ta=n3ta=n4ta=n5ta=0.0

'''START OF EVOLUTION LOOP'''
for k in range(NGEN):

    # --------- P1.3 fitness: Eq. (1.6)/(1.7) ----------
    # lydata columns: [ln q", ln(Tw-Tsat), ln g, ln gamma, ln P]
    for i in range(ND):
        # Build ln term for (g + n4*G_EN*gamma):
        g_val   = ydata[i][2]
        gam_val = ydata[i][3]
        P_val   = ydata[i][4]
        # Safeguards to avoid log of nonpositive due to random exploration:
        mix_g = max(g_val + n[i][4]*G_EN*gam_val, 1e-12)

        Ferr[i]  = n[i][0]*lydata[i][0] + math.log(n[i][1]) + n[i][2]*lydata[i][1]
        Ferr[i] += n[i][3]*math.log(mix_g) + n[i][5]*math.log(P_val)

        aFerr[i] = abs(Ferr[i]) / abs(lydata[i][0])  # absolute fractional error in ln-space
    # ---------------------------------------------------

    aFerrmean = numpy.mean(aFerr)
    meanAFerr[k]=aFerrmean
    aFerrmedian = numpy.median(aFerr)

    '''SELECTION'''
    clim = MFRAC*aFerrmedian
    nkeep = 0
    for j in range(NS):
        if (aFerr[j] < clim):
            nkeep += 1
            ntemp[nkeep-1][1] = n[j][1]
            ntemp[nkeep-1][2] = n[j][2]
            ntemp[nkeep-1][3] = n[j][3]
            ntemp[nkeep-1][4] = n[j][4]
            ntemp[nkeep-1][5] = n[j][5]
    nnew = NS - nkeep  

    '''MATING (now includes n4, n5)'''
    for j in range(nnew): 
        nmate1 = numpy.random.randint(low=0, high=nkeep+1)
        nmate2 = numpy.random.randint(low=0, high=nkeep+1)

        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][1] = n[nmate1][1]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][1] = n[nmate2][1]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][2] = n[nmate1][2]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][2] = n[nmate2][2]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][3] = n[nmate1][3]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][3] = n[nmate2][3]*(1.+0.085*2.*(0.5-numpy.random.rand()))

        # ---- P1.3: include n4, n5 in reproduction ----
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][4] = n[nmate1][4]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][4] = n[nmate2][4]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][5] = n[nmate1][5]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][5] = n[nmate2][5]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        # -----------------------------------------------

    n = deepcopy(ntemp)

    '''POPULATION AVERAGES & MEAN-ERROR USING AVERAGE n'''
    n1avg[k]=n2avg[k]=n3avg[k]=n4avg[k]=n5avg[k]=0.0
    for j in range(NS):
        n1avg[k]+= n[j][1]/NS
        n2avg[k]+= n[j][2]/NS
        n3avg[k]+= n[j][3]/NS
        n4avg[k]+= n[j][4]/NS
        n5avg[k]+= n[j][5]/NS

    for i in range(ND):
        g_val   = ydata[i][2]
        gam_val = ydata[i][3]
        P_val   = ydata[i][4]
        mix_g   = max(g_val + n4avg[k]*G_EN*gam_val, 1e-12)

        Ferravgn[i]  = -1.*lydata[i][0] + math.log(n1avg[k]) + n2avg[k]*lydata[i][1]
        Ferravgn[i] += n3avg[k]*math.log(mix_g) + n5avg[k]*math.log(P_val)

        aFerravgn[i] = abs(Ferravgn[i])/abs(lydata[i][0])
    aFerrmeanavgn[k] = numpy.mean(aFerravgn)

    # time-averages after burn-in
    aFerrta += aFerrmeanavgn[k]/NGEN
    if (k > 800):
        n1ta += n1avg[k]/(NGEN-800)
        n2ta += n2avg[k]/(NGEN-800)
        n3ta += n3avg[k]/(NGEN-800)
        n4ta += n4avg[k]/(NGEN-800)
        n5ta += n5avg[k]/(NGEN-800)

    if (aFerrmeanavgn[k] < aFerrmeanavgnMin):
        aFerrmeanavgnMin = aFerrmeanavgn[k]
        n1min, n2min, n3min, n4min, n5min = n1avg[k], n2avg[k], n3avg[k], n4avg[k], n5avg[k]

# -------- Final printouts (now 5 constants) --------
print('ENDING:  pop.avg  n1..n5,  aFerrmean =',
      n1avg[k], n2avg[k], n3avg[k], n4avg[k], n5avg[k], aFerrmeanavgn[k])
print('MINIMUM: best(avg) n1..n5,  aFerrmeanMin =',
      n1min, n2min, n3min, n4min, n5min, aFerrmeanavgnMin)
print('TIME AVG:   ta(avg) n1..n5,  aFerrmean_ta =',
      n1ta, n2ta, n3ta, n4ta, n5ta, aFerrta)

# ======== Prediction arrays for scatter plot ========
qpppred =  [[0.0]]
qppdata =  [[0.0]]
for i in range(ND-1):
    qpppred.append([0.0]); qppdata.append([0.0])

# ---- P1.3 prediction: Eq. (1.5) using n1min..n5min ----
for i in range(ND):
    TSH = ydata[i][1]
    g_val = ydata[i][2]
    gam   = ydata[i][3]
    P_val = ydata[i][4]
    mix_g = max(g_val + n4min*G_EN*gam, 1e-12)
    qpppred[i] = n1min * (TSH**n2min) * (mix_g**n3min) * (P_val**n5min)
    qppdata[i] = ydata[i][0]

# ======== Plots ========
plt.rcParams.update({'font.size': 18})

# evolution of error and constants
plt.plot(gen, aFerrmeanavgn)
plt.plot(gen, n1avg)
plt.plot(gen, n2avg)
plt.plot(gen, n3avg)
plt.plot(gen, n4avg)
plt.plot(gen, n5avg)
plt.legend(['aFerrmeanavgn','n1 avg','n2 avg','n3 avg','n4 avg','n5 avg'], loc='upper right')
plt.xlabel('generation'); plt.ylabel('constants and error')
plt.loglog(); plt.yticks([0.01,0.1,1.0,10]); plt.xticks([1,10,100,1000,10000])
plt.show()

# data vs prediction scatter (log-log)
plt.scatter(qpppred, qppdata)
plt.title('Genetic Algorithm (P1.3 five-constant model)')
plt.xlabel('predicted heat flux (W/cm^2), n1..n5 best-fit')
plt.ylabel('measured heat flux (W/cm^2)')
plt.loglog(); plt.xlim(10, 1000); plt.ylim(10, 1000)
plt.show()

# ======== Surface plot of q''/(ΔT)^{n2} vs g and γ at P = 10 kPa ========

# Pressure fixed at 10 kPa
P_fixed = 10.0  

# Generate grid for g (0–20 m/s²) and gamma (0–2)
g_vals = np.linspace(0.0, 20.0, 81)
gamma_vals = np.linspace(0.0, 2.0, 81)
G, GAM = np.meshgrid(g_vals, gamma_vals, indexing='xy')

# Use best-fit constants from GA result (n1min..n5min)
# q''/(ΔT)^{n2} = n1 * (g + n4*G_EN*γ)^{n3} * (P)^{n5}
mix_g = np.maximum(G + n4min * G_EN * GAM, 1e-12)
Z = n1min * (mix_g ** n3min) * (P_fixed ** n5min)

# 3D Surface Plot
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(G, GAM, Z, cmap='viridis', edgecolor='none')
ax.set_xlabel('g (m/s²)')
ax.set_ylabel('γ (surface tension parameter)')
ax.set_zlabel(r"$q''/(T_w - T_{sat})^{n_2}$")
ax.set_title("Surface: q''/(ΔT)^{n2} vs g and γ @ P = 10 kPa")
fig.colorbar(surf, shrink=0.5, aspect=10, label='q\'\'/(ΔT)^{n2}')
plt.tight_layout()
plt.show()

```
