```python
'''>>>>> start CodeP1.2F25
    V.P. Carey ME249, Fall 2025''' 
from random import random

'''INITIALIZING PARAMETERS'''
n = []
ntemp = []
gen=[0]
n1avg = [0.0]
n2avg = [0.0]
n3avg = [0.0]
n4avg = [0.0]
n5avg = [0.0]
meanAFerr=[0.0]
aFerrmeanavgn=[0.0] 

#set program parameters
NGEN = 6000      #number of generations (steps)
MFRAC = 0.5   # faction of median threshold

# here the number of data vectors equals the number of DNA strands (or organisms)
# they can be different if they are randomly paired to compute Ferr (survivability) function
for k in range(NGEN-1):
    gen.append(k+1)   # generation array stores the
    meanAFerr.append(0.0)
    aFerrmeanavgn.append(0.0)
    n1avg.append(0.0)
    n2avg.append(0.0)
    n3avg.append(0.0)
    n4avg.append(0.0)
    n5avg.append(0.0)
    
'''guesses for initial solution population'''
n0i =  -1.0 
n1i = 0.00027
n2i = 4.0
n3i = 0.063
n4i = 1.215 
n5i = 0.145

"3 extra Runs"
# Set 2: Moderate variation
n1i = 0.0005
n2i = 3.5
n3i = 0.08

# Set 3: Larger variation  
n1i = 0.001
n2i = 3.0
n3i = 0.1

# # Set 4: Different direction
n1i = 0.0005
n2i = 3.3
n3i = 0.056



#- initialize arrays before start of evolution loop EL
#then - create array of DNA strands n[i] and ntemp[i] with dimesnion NS = 5

#i initialize array where rows are dna vectors [n0i,n1i,...n5i] with random pertubations
n =  [[-1., n1i+0.001*random(), n2i+0.1*random(), n3i+0.0001*random(), n4i+0.0001*random(), n5i+0.0001*random()]]
for i in range(ND):
    n.append([-1., n1i+0.0001*random(), n2i+0.001*random(), n3i+0.0001*random(), n4i+0.0001*random(), n5i+0.0001*random()])
#print (n) # uncomment command to print array so it can be checked

# store also in wtemp
ntemp = deepcopy(n)

#initialize Ferr values an dother loop parameters
#define arrays of Ferr (error) functions
#individual solution error and absoute error
Ferr =  [[0.0]]  
#population average solution error and absoute error
Ferravgn =  [[0.0]]
aFerr =  [[0.0]]
aFerravgn =  [[0.0]]

#store zeros in ND genes
for i in range(ND-1):
    #individual solution error and absoute error
    Ferr.append([0.0])
    aFerr.append([0.0])
    #population average solution error and absoute error
    Ferravgn.append([0.0])
    aFerravgn.append([0.0])
#print (Ferr)

aFerrmeanavgnMin=1000000000.0 
# these store the  n values for minimum population average error durng NGEN generations
n1min = 0.0
n2min = 0.0
n3min = 0.0
n4min = 0.0
n5min = 0.0
aFerrta = 0.0
# these store the time averaged n values durng from generation 800 to NGEN generations
n1min = 0.0
n1ta = 0.0
n2ta = 0.0
n3ta = 0.0
n4ta = 0.0
n5ta = 0.0

'''START OF EVOLUTION LOOP'''
# ------------------------------------------
# k is generation number, NGEN IS TOTAL NUMBER OF GENERATIONS COMPUTED
for k in range(NGEN):
    
    '''In this program , the number of organisms (solutions) NS is taken to be equal to the 
    number of data points ND so for each generation, each solution can be compared to a different 
    data point and all the data is compared in each generation.  The order of the solutions in the array 
    that holds the solution constants is constantly changing due to mating and selection, so the pairing 
    is random.'''
    
    
    '''CALCULATING ERROR (FITNESS)
    In this program, the absolute error in the logrithm of the physical heat flux is 
    used to evaluate fitness.'''
    
    # Here we calculate error Ferr and absolute error aFerr for each data point and organism pair 
    # for specified n(i), and calculate (mean aFerr) = aFerrmean
    # and (median aFerr) = aFerrmedian for the data collection and specified n(i)
    # Note that the number data points ND equals the number of solutions (organisms) in the population
    #====================================================================================
    '''CALCULATING ERROR (FITNESS)'''
    for i in range(ND):
        
        Ferr[i] = n[i][0]*lydata[i][0] + math.log(n[i][1]) + n[i][2]*lydata[i][1]
        Ferr[i] = Ferr[i] + n[i][3]*math.log( ydata[i][2] ) 
           
        aFerr[i] = abs(Ferr[i])/abs(lydata[i][0])  #- absolute fractional error
    #-------------
    aFerrmean = numpy.mean(aFerr) #mean error for population for this generation
    meanAFerr[k]=aFerrmean  #store aFerrmean for this generation gen[k]=k
    aFerrmedian = numpy.median(aFerr) #median error for population for this generation

    '''SELECTION'''
    #pick survivors 
    #[2] calculate survival cutoff, set number kept = nkeep = 0
    #==============================================
    clim = MFRAC*aFerrmedian  #cut off limit is a fraction/multiplier MFRAC of median value for population
    nkeep = 0
    
    # now check each organism/solution to see if aFerr is less than cut of limit
    #if yes, store n for next generation population in ntemp, at end nkeep = number that survive
    #and number of new offspring = NS-nkeep
    #==============================================
    for j in range(NS):  # NS Ferr values, one for each solution in population
        if (aFerr[j] < clim):
            nkeep = nkeep + 1
            #ntemp[nkeep][0] = n[j][0] = -1 so it is unchanged;
            ntemp[nkeep-1][1] = n[j][1];
            ntemp[nkeep-1][2] = n[j][2];
            ntemp[nkeep-1][3] = n[j][3];
            ntemp[nkeep-1][4] = n[j][4];
            ntemp[nkeep-1][5] = n[j][5];
    #now have survivors in leading entries in list of ntemp vectors from 1 to nkeep
    #compute number to be added by mating
    nnew = NS - nkeep  
    
    '''MATING'''
    #[4] for nnew new organisms/solutions,
    # randomly pick two survivors, randomly pick DNA (n) from pair for each offspring
    #==============================================================
    for j in range(nnew): 
        # pick two survivors randomly
        nmate1 = numpy.random.randint(low=0, high=nkeep+1)
        nmate2 = numpy.random.randint(low=0, high=nkeep+1)
        
        #then randomly pick DNA from parents for offspring
        
        '''here, do not change property ntemp[nkeep+j+1][0], it's always fixed at the value -1'''
        #if (numpy.random.rand() < 0.5)
        #    ntemp[nkeep+j+1][0] = n[nmate1][0]  
        #else
        #    ntemp[nkeep+j+1][0] = n[nmate2][0]
      
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][1] = n[nmate1][1]*(1.+0.085*2.*(0.5-numpy.random.rand()))  # property 1, mutation added
        else:
            ntemp[nkeep+j+1][1] = n[nmate2][1]*(1.+0.085*2.*(0.5-numpy.random.rand()))
    
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][2] = n[nmate1][2]*(1.+0.085*2.*(0.5-numpy.random.rand()))  # property 2, mutation added
        else:
            ntemp[nkeep+j+1][2] = n[nmate2][2]*(1.+0.085*2.*(0.5-numpy.random.rand()))
    
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][3] = n[nmate1][3]*(1.+0.085*2.*(0.5-numpy.random.rand()))  # property 3, mutation added
        else:
            ntemp[nkeep+j+1][3] = n[nmate2][3]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        '''    
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][4] = n[nmate1][4]*(1.+0.085*2.*(0.5-numpy.random.rand()))  # property 4, mutation added
        else:
            ntemp[nkeep+j+1][4] = n[nmate2][4]*(1.+0.085*2.*(0.5-numpy.random.rand()))
                
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][5] = n[nmate1][5]*(1.+0.085*2.*(0.5-numpy.random.rand()))  # property 5, mutation added
        else:
            ntemp[nkeep+j+1][5] = n[nmate2][5]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        '''
    #=============================================
    n = deepcopy(ntemp)   # save ntemp as n for use in next generation (next iteration)
    
    
    '''AVERAGING OVER POPULATION AND OVER TIME, FINDING MINIMUM ERROR SET OF n VALUES'''
    # [6] calculate n1avg[k], etc., which are average n values for population of organisms (solutions) 
    # at this generation k
    #=============================================
    #initialize average n's to zero and sum contribution of each member of the solution population 
    n1avg[k] = 0.0;
    n2avg[k] = 0.0;
    n3avg[k] = 0.0;
    n4avg[k] = 0.0;
    n5avg[k] = 0.0;
    for j in range(NS):    
        n1avg[k] = n1avg[k] + n[j][1]/NS; 
        n2avg[k] = n2avg[k] + n[j][2]/NS; 
        n3avg[k] = n3avg[k] + n[j][3]/NS;
        n4avg[k] = n4avg[k] + n[j][4]/NS;
        n5avg[k] = n5avg[k] + n[j][5]/NS;

    # Here we compute aFerravgn[i] = absolute Ferr of logrithm data point i using average n values 
    # for this solutions generation k
    # aFerrmeanavgn[k] is the mean of the Ferravgn[i] for the population of organisms/solutions
    # 
    #=============================================
    ''' CALCULATING MEAN ERROR FOR POPULATION'''
    for i in range(ND): 
        Ferravgn[i] = -1.*lydata[i][0] + math.log(n1avg[k]) + n2avg[k]*lydata[i][1]
        Ferravgn[i] = Ferravgn[i] + n3avg[k]*math.log( ydata[i][2] ) 
            
        #aFerravgn[i] = abs(Ferr[i])/abs(lydata[i][0])
        aFerravgn[i] = abs(Ferravgn[i])/abs(lydata[i][0])
    #-------------
    aFerrmeanavgn[k] = numpy.mean(aFerravgn)
    
    
    # next, update time average of n valaues in population (n1ta[k], etc.) 
    # for generations = k > 800 up to total NGEN
    #=============================================
    aFerrta = aFerrta + aFerrmeanavgn[k]/NGEN
    if (k > 800):
        n1ta = n1ta + n1avg[k]/(NGEN-800)
        n2ta = n2ta + n2avg[k]/(NGEN-800)
        n3ta = n3ta + n3avg[k]/(NGEN-800)
        n4ta = n4ta + n4avg[k]/(NGEN-800)
        n5ta = n5ta + n5avg[k]/(NGEN-800)
    
    # compare aFerrmeanavgn[k] to previous minimum value and save 
    # it and corresponding n(i) values if the value for this generation k is lower 
    #=============================================
    if (aFerrmeanavgn[k] < aFerrmeanavgnMin):
        aFerrmeanavgnMin = aFerrmeanavgn[k] 
        n1min = n1avg[k]
        n2min = n2avg[k]
        n3min = n3avg[k]
        n4min = n4avg[k]
        n5min = n5avg[k]
    
    #print('avg n1-n4:', n1avg[k], n2avg[k], n3avg[k], n4avg[k], aFerrmeanavgn[k])
    #print ('kvalue =', k)
    '''end of evolution loop'''
    # ------------------------------------------------------------------
    # ------------------------------------------------------------------
    
    
# ------------------------------------------------------------------
#final print and plot of results
# ------------------------------------------------------------------
print('ENDING: pop. avg n1-n3,aFerrmean:', n1avg[k], n2avg[k], n3avg[k], aFerrmeanavgn[k])
print('MINUMUM:  avg n1-n3,aFerrmeanMin:', n1min, n2min, n3min, aFerrmeanavgnMin)
print('TIME AVG:    avg n1-n3,aFerrmean:', n1ta, n2ta, n3ta, aFerrta)

#SETTING UP PLOTS

#========
#initialize values
qpppred =  [[0.0]]
qppdata =  [[0.0]]
for i in range(ND-1):
    qpppred.append([0.0])
    qppdata.append([0.0])   
#calculate predicted and data values to plot
for i in range(ND):
    qpppred[i] = n1min*(ydata[i][1]**n2min) * ((ydata[i][2])**n3min) 
    qppdata[i] = ydata[i][0] 
    
#========

# constants evolution plots
# x axis values are generation number 
# corresponding y axis values are mean absolute population error aFerrmeanavgn and n values
# plotting the points  

plt.rcParams.update({'font.size': 18})

# aFerrmeanavgn[k] is the mean of the Ferravgn[i] for the population of organisms/solutions
# computed using the mean n values
plt.plot(gen, aFerrmeanavgn) 
plt.plot(gen, n1avg)
plt.plot(gen, n2avg)
plt.plot(gen, n3avg)
plt.legend(['aFerrmeanavgn', 'n1 avg', 'n2 avg', 'n3 avg'], loc='lower left')
#plt.plot(gen, n4avg)
#plt.plot(gen, n5avg)
#plt.legend(['aFerrmeanavgn', 'n1 avg', 'n2 avg', 'n3 avg', 'n4 avg', 'n5 avg'], loc='upper right')

# naming the x axis 
plt.xlabel('generation') 
# naming the y axis 
plt.ylabel('constants and error') 
plt.loglog()
plt.yticks([0.01,0.1,1.0,10])
plt.xticks([1,10,100,1000,10000]) 
plt.savefig('task1_2_constantsanderror.png', dpi=300, bbox_inches='tight')

plt.show() 


'''CALCULATE PREDICTED VALUES AND RETRIEVE DATA VALUES TO PLOT'''
plt.scatter(qpppred, qppdata)
plt.title('Genetic Algorithm')
plt.xlabel('predicted heat flux (W/cm^2), n1min - n3min constants')
plt.ylabel('measured heat flux (W/cm^2)')
plt.loglog()
plt.xlim(xmax = 1000, xmin = 10)
plt.ylim(ymax = 1000, ymin = 10)
plt.savefig('task1_2_geneticalgorithim.png', dpi=300, bbox_inches='tight')

plt.show()
    
# end CodeP1.2F25 


```

```python
'''>>>>> start CodeP1.2F25 (P1.3-modified: 5-constant model with γ and P)
    V.P. Carey ME249, Fall 2025'''

from random import random
import math, numpy
from copy import deepcopy
import matplotlib.pyplot as plt
import numpy as np
# ---------- P1.3 additions ----------
G_EN = 9.8  # earth-normal gravity (m/s^2) used in g + n4*G_EN*gamma
# -----------------------------------

'''INITIALIZING PARAMETERS'''
n = []
ntemp = []
gen=[0]
n1avg = [0.0]
n2avg = [0.0]
n3avg = [0.0]
n4avg = [0.0]
n5avg = [0.0]
meanAFerr=[0.0]
aFerrmeanavgn=[0.0] 

#set program parameters
NGEN = 6000      #number of generations (steps)
MFRAC = 0.5      # fraction of median threshold

# Build storage arrays
for k in range(NGEN-1):
    gen.append(k+1)
    meanAFerr.append(0.0)
    aFerrmeanavgn.append(0.0)
    n1avg.append(0.0)
    n2avg.append(0.0)
    n3avg.append(0.0)
    n4avg.append(0.0)
    n5avg.append(0.0)

# ---------- P1.3 initial guesses (per handout) ----------
n0i =  -1.0  
n1i = 0.000476 
n2i = 3.028 
n3i = 0.2249 
n4i = 1.054  
n5i = 0.217
# --------------------------------------------------------

# Create initial population (NS rows) with small random perturbations
# NOTE: n[i][0] is fixed at -1 to carry the sign used in Ferr; others evolve
n =  [[-1., n1i+0.001*random(), n2i+0.1*random(), n3i+0.0001*random(), n4i+0.0001*random(), n5i+0.0001*random()]]
for i in range(NS):  # NS == ND in this program
    n.append([-1., n1i+0.0001*random(), n2i+0.001*random(), n3i+0.0001*random(), n4i+0.0001*random(), n5i+0.0001*random()])

ntemp = deepcopy(n)

# Error arrays
Ferr =  [[0.0]]  
Ferravgn =  [[0.0]]
aFerr =  [[0.0]]
aFerravgn =  [[0.0]]
for i in range(ND-1):
    Ferr.append([0.0]); aFerr.append([0.0])
    Ferravgn.append([0.0]); aFerravgn.append([0.0])

aFerrmeanavgnMin=1e30

# Trackers
n1min=n2min=n3min=n4min=n5min=0.0
aFerrta=0.0
n1ta=n2ta=n3ta=n4ta=n5ta=0.0

'''START OF EVOLUTION LOOP'''
for k in range(NGEN):

    # --------- P1.3 fitness: Eq. (1.6)/(1.7) ----------
    # lydata columns: [ln q", ln(Tw-Tsat), ln g, ln gamma, ln P]
    for i in range(ND):
        # Build ln term for (g + n4*G_EN*gamma):
        g_val   = ydata[i][2]
        gam_val = ydata[i][3]
        P_val   = ydata[i][4]
        # Safeguards to avoid log of nonpositive due to random exploration:
        mix_g = max(g_val + n[i][4]*G_EN*gam_val, 1e-12)

        Ferr[i]  = n[i][0]*lydata[i][0] + math.log(n[i][1]) + n[i][2]*lydata[i][1]
        Ferr[i] += n[i][3]*math.log(mix_g) + n[i][5]*math.log(P_val)

        aFerr[i] = abs(Ferr[i]) / abs(lydata[i][0])  # absolute fractional error in ln-space
    # ---------------------------------------------------

    aFerrmean = numpy.mean(aFerr)
    meanAFerr[k]=aFerrmean
    aFerrmedian = numpy.median(aFerr)

    '''SELECTION'''
    clim = MFRAC*aFerrmedian
    nkeep = 0
    for j in range(NS):
        if (aFerr[j] < clim):
            nkeep += 1
            ntemp[nkeep-1][1] = n[j][1]
            ntemp[nkeep-1][2] = n[j][2]
            ntemp[nkeep-1][3] = n[j][3]
            ntemp[nkeep-1][4] = n[j][4]
            ntemp[nkeep-1][5] = n[j][5]
    nnew = NS - nkeep  

    '''MATING (now includes n4, n5)'''
    for j in range(nnew): 
        nmate1 = numpy.random.randint(low=0, high=nkeep+1)
        nmate2 = numpy.random.randint(low=0, high=nkeep+1)

        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][1] = n[nmate1][1]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][1] = n[nmate2][1]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][2] = n[nmate1][2]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][2] = n[nmate2][2]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][3] = n[nmate1][3]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][3] = n[nmate2][3]*(1.+0.085*2.*(0.5-numpy.random.rand()))

        # ---- P1.3: include n4, n5 in reproduction ----
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][4] = n[nmate1][4]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][4] = n[nmate2][4]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        if (numpy.random.rand() < 0.5):
            ntemp[nkeep+j+1][5] = n[nmate1][5]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        else:
            ntemp[nkeep+j+1][5] = n[nmate2][5]*(1.+0.085*2.*(0.5-numpy.random.rand()))
        # -----------------------------------------------

    n = deepcopy(ntemp)

    '''POPULATION AVERAGES & MEAN-ERROR USING AVERAGE n'''
    n1avg[k]=n2avg[k]=n3avg[k]=n4avg[k]=n5avg[k]=0.0
    for j in range(NS):
        n1avg[k]+= n[j][1]/NS
        n2avg[k]+= n[j][2]/NS
        n3avg[k]+= n[j][3]/NS
        n4avg[k]+= n[j][4]/NS
        n5avg[k]+= n[j][5]/NS

    for i in range(ND):
        g_val   = ydata[i][2]
        gam_val = ydata[i][3]
        P_val   = ydata[i][4]
        mix_g   = max(g_val + n4avg[k]*G_EN*gam_val, 1e-12)

        Ferravgn[i]  = -1.*lydata[i][0] + math.log(n1avg[k]) + n2avg[k]*lydata[i][1]
        Ferravgn[i] += n3avg[k]*math.log(mix_g) + n5avg[k]*math.log(P_val)

        aFerravgn[i] = abs(Ferravgn[i])/abs(lydata[i][0])
    aFerrmeanavgn[k] = numpy.mean(aFerravgn)

    # time-averages after burn-in
    aFerrta += aFerrmeanavgn[k]/NGEN
    if (k > 800):
        n1ta += n1avg[k]/(NGEN-800)
        n2ta += n2avg[k]/(NGEN-800)
        n3ta += n3avg[k]/(NGEN-800)
        n4ta += n4avg[k]/(NGEN-800)
        n5ta += n5avg[k]/(NGEN-800)

    if (aFerrmeanavgn[k] < aFerrmeanavgnMin):
        aFerrmeanavgnMin = aFerrmeanavgn[k]
        n1min, n2min, n3min, n4min, n5min = n1avg[k], n2avg[k], n3avg[k], n4avg[k], n5avg[k]

# -------- Final printouts (now 5 constants) --------
print('ENDING:  pop.avg  n1..n5,  aFerrmean =',
      n1avg[k], n2avg[k], n3avg[k], n4avg[k], n5avg[k], aFerrmeanavgn[k])
print('MINIMUM: best(avg) n1..n5,  aFerrmeanMin =',
      n1min, n2min, n3min, n4min, n5min, aFerrmeanavgnMin)
print('TIME AVG:   ta(avg) n1..n5,  aFerrmean_ta =',
      n1ta, n2ta, n3ta, n4ta, n5ta, aFerrta)

# ======== Prediction arrays for scatter plot ========
qpppred =  [[0.0]]
qppdata =  [[0.0]]
for i in range(ND-1):
    qpppred.append([0.0]); qppdata.append([0.0])

# ---- P1.3 prediction: Eq. (1.5) using n1min..n5min ----
for i in range(ND):
    TSH = ydata[i][1]
    g_val = ydata[i][2]
    gam   = ydata[i][3]
    P_val = ydata[i][4]
    mix_g = max(g_val + n4min*G_EN*gam, 1e-12)
    qpppred[i] = n1min * (TSH**n2min) * (mix_g**n3min) * (P_val**n5min)
    qppdata[i] = ydata[i][0]

# ======== Plots ========
plt.rcParams.update({'font.size': 18})

# evolution of error and constants
plt.plot(gen, aFerrmeanavgn)
plt.plot(gen, n1avg)
plt.plot(gen, n2avg)
plt.plot(gen, n3avg)
plt.plot(gen, n4avg)
plt.plot(gen, n5avg)
plt.legend(['aFerrmeanavgn','n1 avg','n2 avg','n3 avg','n4 avg','n5 avg'], loc='upper right')
plt.xlabel('generation'); plt.ylabel('constants and error')
plt.loglog(); plt.yticks([0.01,0.1,1.0,10]); plt.xticks([1,10,100,1000,10000])
plt.show()

# data vs prediction scatter (log-log)
plt.scatter(qpppred, qppdata)
plt.title('Genetic Algorithm (P1.3 five-constant model)')
plt.xlabel('predicted heat flux (W/cm^2), n1..n5 best-fit')
plt.ylabel('measured heat flux (W/cm^2)')
plt.loglog(); plt.xlim(10, 1000); plt.ylim(10, 1000)
plt.show()

## ======== Surface plot of q''/(ΔT)^{n2} vs g and γ at P = 10 kPa ========
# Pressure fixed at 10 kPa
P_fixed = 10.0  

# Generate grid for g (0–20 m/s²) and gamma (0–2)
g_vals = np.linspace(0.0, 20.0, 100)
gamma_vals = np.linspace(0.0, 2.0, 100)
G, GAM = np.meshgrid(g_vals, gamma_vals, indexing='xy')

# Use best-fit constants from GA result (n1min..n5min)
# q''/(ΔT)^{n2} = n1 * (g + n4*G_EN*γ)^{n3} * (P)^{n5}
mix_g = np.maximum(G + n4min * G_EN * GAM, 1e-12)
Z = n1min * (mix_g ** n3min) * (P_fixed ** n5min)

# 3D Surface Plot
fig = plt.figure(figsize=(14, 10))
ax = fig.add_subplot(111, projection='3d')

# Create surface with improved styling
surf = ax.plot_surface(G, GAM, Z, cmap='viridis', 
                       edgecolor='none', alpha=0.9,
                       linewidth=0, antialiased=True)

# Enhanced labels with LaTeX formatting
ax.set_xlabel('g (m/s²)', fontsize=16, labelpad=15)
ax.set_ylabel('γ (surface tension parameter)', fontsize=16, labelpad=15)
ax.set_zlabel(r"$q''/\Delta T^{n_2}$", fontsize=16, labelpad=15)
ax.set_title(f"Heat Flux Correlation Surface at P = {P_fixed} kPa",
             fontsize=18, pad=25)

# Adjust viewing angle for better perspective
ax.view_init(elev=25, azim=45)

# Add colorbar with better formatting
cbar = fig.colorbar(surf, ax=ax, shrink=0.5, aspect=12, pad=0.15)
cbar.set_label(r"$q''/\Delta T^{n_2}$ (W/cm²·K$^{-n_2}$)", 
               fontsize=14, rotation=270, labelpad=30)

# Add grid for better readability
ax.grid(True, alpha=0.3, linewidth=0.5)

# Better tick spacing (fewer ticks for clarity)
ax.set_xticks([0, 5, 10, 15, 20])
ax.set_yticks([0, 0.5, 1.0, 1.5, 2.0])
ax.zaxis.set_major_locator(plt.MaxNLocator(5))

# Adjust tick label size
ax.tick_params(axis='both', which='major', labelsize=12)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt
from random import random

# Extract data columns
qdata_train = training[:, 2]  # Heat transfer rate (W/cm²)
T_a_in_train = training[:, 0]  # T_a,in (°C)
T_a_out_train = training[:, 1]  # T_a,out (°C)

qdata_val = validation[:, 2]
T_a_in_val = validation[:, 0]
T_a_out_val = validation[:, 1]

# Physical constants
sigma = 5.67e-8  # Stefan-Boltzmann constant (W/m²K⁴)
h_c_ref = 80.0  # Reference convection coefficient (W/m²K) from project

# Number of training data points (organisms)
ND = len(training)
NS = ND  # Number of solutions equals number of data points

'''INITIALIZING PARAMETERS'''
n = []
ntemp = []
gen = [0]
n1avg = [0.0]  # (UA)_e average
n2avg = [0.0]  # (UA)_c average
n3avg = [0.0]  # α average
meanAFerr = [0.0]
aFerrmeanavgn = [0.0]

# Set program parameters
NGEN = 5000  # Number of generations
MFRAC = 0.5  # Fraction of median threshold

# Initialize generation arrays
for k in range(NGEN-1):
    gen.append(k+1)
    meanAFerr.append(0.0)
    aFerrmeanavgn.append(0.0)
    n1avg.append(0.0)
    n2avg.append(0.0)
    n3avg.append(0.0)

'''Initial guesses for solution population'''
# n1 = (UA)_e, n2 = (UA)_c, n3 = α
n1i = 100.0  # Initial guess for (UA)_e
n2i = 100.0   # Initial guess for (UA)_c
n3i = 0.4  # Initial guess for α

# Initialize array of DNA strands with random perturbations
n = [[n1i + 0.2*random(), n2i + 0.2*random(), n3i + 0.1*random()]]
for i in range(ND-1):
    n.append([n1i + 0.2*random(), n2i + 0.2*random(), n3i + 0.1*random()])

# Store also in ntemp
ntemp = deepcopy(n)

# Initialize error arrays
Ferr = [[0.0] for _ in range(ND)]
Ferravgn = [[0.0] for _ in range(ND)]
aFerr = [[0.0] for _ in range(ND)]
aFerravgn = [[0.0] for _ in range(ND)]

# Initialize tracking variables
aFerrmeanavgnMin = 1000000000.0
n1min = 0.0
n2min = 0.0
n3min = 0.0
aFerrta = 0.0
n1ta = 0.0
n2ta = 0.0
n3ta = 0.0

# Validation error tracking
val_errors = []

'''FUNCTION TO CALCULATE HEAT TRANSFER'''
def calculate_heat_transfer(n1, n2, n3, T_in, T_out):
    """Calculate predicted heat transfer using equation 2.3/2.6 from project"""
    # Convert temperatures to Kelvin for radiation calculation
    T_in_K = T_in + 273.15
    T_out_K = T_out + 273.15
    
    # Calculate h_r = σ(T_a,in + T_a,out)(T²_a,in + T²_a,out)
    h_r = sigma * (T_in_K + T_out_K) * (T_in_K**2 + T_out_K**2)
    
    # Calculate heat transfer using equation 2.6
    # q̇ = n2 * (1 + n3*h_r/h_c_ref) * [n1(T_a,in - T_a,out)] / [n1 + n2(1 + n3*h_r/h_c_ref)]
    
    numerator = n1 * (T_in - T_out)
    denominator = n1 + n2 * (1 + n3 * h_r / h_c_ref)
    
    # Avoid division by zero
    if abs(denominator) < 1e-10:
        return 0.0
    
    q_pred = n2 * (1 + n3 * h_r / h_c_ref) * numerator / denominator
    
    return q_pred

'''START OF EVOLUTION LOOP'''
for k in range(NGEN):
    
    '''CALCULATING ERROR (FITNESS) using equation 2.4'''
    for i in range(ND):
        # Calculate predicted heat transfer
        q_pred = calculate_heat_transfer(n[i][0], n[i][1], n[i][2], 
                                        T_a_in_train[i], T_a_out_train[i])
        
        # Calculate error (equation 2.4): ferr,i = qdata,i - q_pred
        Ferr[i] = qdata_train[i] - q_pred
        
        # Calculate fractional absolute error
        aFerr[i] = abs(Ferr[i]) / abs(qdata_train[i])
    
    # Calculate population statistics
    aFerrmean = np.mean(aFerr)
    meanAFerr[k] = aFerrmean
    aFerrmedian = np.median(aFerr)
    
    '''SELECTION'''
    clim = MFRAC * aFerrmedian  # Cutoff limit
    nkeep = 0
    
    # Select survivors
    for j in range(NS):
        if aFerr[j] < clim:
            nkeep = nkeep + 1
            ntemp[nkeep-1][0] = n[j][0]  # (UA)_e
            ntemp[nkeep-1][1] = n[j][1]  # (UA)_c
            ntemp[nkeep-1][2] = n[j][2]  # α
    
    # Handle case where no organisms survive
    if nkeep == 0:
        nkeep = 1
        ntemp[0][0] = n[np.argmin(aFerr)][0]
        ntemp[0][1] = n[np.argmin(aFerr)][1]
        ntemp[0][2] = n[np.argmin(aFerr)][2]
    
    nnew = NS - nkeep  # Number of offspring needed
    
    '''MATING with mutation'''
    for j in range(nnew):
        # Pick two survivors randomly
        nmate1 = np.random.randint(low=0, high=nkeep)
        nmate2 = np.random.randint(low=0, high=nkeep)
        
        # Create offspring with mutation for (UA)_e
        if np.random.rand() < 0.5:
            ntemp[nkeep+j][0] = n[nmate1][0] * (1 + 0.09 * 2 * (0.5 - np.random.rand()))
        else:
            ntemp[nkeep+j][0] = n[nmate2][0] * (1 + 0.09 * 2 * (0.5 - np.random.rand()))
        
        # Ensure positive values
        ntemp[nkeep+j][0] = max(ntemp[nkeep+j][0], 0.01)
        
        # Create offspring with mutation for (UA)_c
        if np.random.rand() < 0.5:
            ntemp[nkeep+j][1] = n[nmate1][1] * (1 + 0.09 * 2 * (0.5 - np.random.rand()))
        else:
            ntemp[nkeep+j][1] = n[nmate2][1] * (1 + 0.09 * 2 * (0.5 - np.random.rand()))
        
        ntemp[nkeep+j][1] = max(ntemp[nkeep+j][1], 0.01)
        
        # Create offspring with mutation for α
        if np.random.rand() < 0.5:
            ntemp[nkeep+j][2] = n[nmate1][2] * (1 + 0.09 * 2 * (0.5 - np.random.rand()))
        else:
            ntemp[nkeep+j][2] = n[nmate2][2] * (1 + 0.09 * 2 * (0.5 - np.random.rand()))
        
        ntemp[nkeep+j][2] = max(ntemp[nkeep+j][2], 0.0)
    
    # Update population
    n = deepcopy(ntemp)
    
    '''CALCULATE POPULATION AVERAGES'''
    n1avg[k] = 0.0
    n2avg[k] = 0.0
    n3avg[k] = 0.0
    
    for j in range(NS):
        n1avg[k] += n[j][0] / NS
        n2avg[k] += n[j][1] / NS
        n3avg[k] += n[j][2] / NS
    
    '''CALCULATE MEAN ERROR FOR POPULATION AVERAGE'''
    for i in range(ND):
        q_pred_avg = calculate_heat_transfer(n1avg[k], n2avg[k], n3avg[k],
                                            T_a_in_train[i], T_a_out_train[i])
        Ferravgn[i] = qdata_train[i] - q_pred_avg
        aFerravgn[i] = abs(Ferravgn[i])
    
    aFerrmeanavgn[k] = np.mean(aFerravgn)
    
    '''CALCULATE VALIDATION ERROR'''
    val_error = 0.0
    for i in range(len(validation)):
        q_pred_val = calculate_heat_transfer(n1avg[k], n2avg[k], n3avg[k],
                                            T_a_in_val[i], T_a_out_val[i])
        error_val = qdata_val[i] - q_pred_val
        if abs(qdata_val[i]) > 1e-10:
            val_error += abs(error_val) / abs(qdata_val[i])
        else:
            val_error += abs(error_val)
    val_error = val_error / len(validation)
    val_errors.append(val_error)
    
    '''TIME AVERAGING (after generation 800)'''
    aFerrta += aFerrmeanavgn[k] / NGEN
    if k > 800:
        n1ta += n1avg[k] / (NGEN - 800)
        n2ta += n2avg[k] / (NGEN - 800)
        n3ta += n3avg[k] / (NGEN - 800)
    
    '''UPDATE MINIMUM'''
    if aFerrmeanavgn[k] < aFerrmeanavgnMin:
        aFerrmeanavgnMin = aFerrmeanavgn[k]
        n1min = n1avg[k]
        n2min = n2avg[k]
        n3min = n3avg[k]

'''END OF EVOLUTION LOOP'''

# Final results
print('\n=== FINAL RESULTS ===')
print(f'ENDING: pop. avg (UA)e={n1avg[-1]:.4f}, (UA)c={n2avg[-1]:.4f}, α={n3avg[-1]:.4f}, Error={aFerrmeanavgn[-1]:.4f}')
print(f'MINIMUM: (UA)e={n1min:.4f}, (UA)c={n2min:.4f}, α={n3min:.4f}, Error={aFerrmeanavgnMin:.4f}')
print(f'TIME AVG: (UA)e={n1ta:.4f}, (UA)c={n2ta:.4f}, α={n3ta:.4f}, Error={aFerrta:.4f}')
print(f'Final Training Error: {aFerrmeanavgn[-1]:.4f}')
print(f'Final Validation Error: {val_errors[-1]:.4f}')

'''PLOTTING RESULTS'''
plt.rcParams.update({'font.size': 12})

# Plot 1: Evolution of parameters and error
fig = plt.figure(figsize=(16, 6))

plt.subplot(1, 3, 1)
plt.plot(gen, aFerrmeanavgn, 'b-', linewidth=2, label='Training Error')
plt.plot(gen, val_errors, 'r--', linewidth=2, label='Validation Error')
plt.legend(loc='upper right')
plt.xlabel('Generation')
plt.ylabel('Mean Absolute Fractional Error')
plt.loglog()
plt.grid(True, alpha=0.3)
plt.title('Error Evolution')

plt.subplot(1, 3, 2)
plt.plot(gen, n1avg, 'b-', linewidth=2, label='(UA)_e')
plt.plot(gen, n2avg, 'r-', linewidth=2, label='(UA)_c')
plt.plot(gen, n3avg, 'g-', linewidth=2, label='α')
plt.legend(loc='best')
plt.xlabel('Generation')
plt.ylabel('Parameter Values')
plt.semilogx()
plt.grid(True, alpha=0.3)
plt.title('Parameter Evolution')

# Plot 2: Predicted vs Measured Heat Transfer
plt.subplot(1, 3, 3)

# Calculate predictions using best parameters
qpred_train = []
qpred_val = []

for i in range(ND):
    q_pred = calculate_heat_transfer(n1min, n2min, n3min,
                                    T_a_in_train[i], T_a_out_train[i])
    qpred_train.append(q_pred)

for i in range(len(validation)):
    q_pred = calculate_heat_transfer(n1min, n2min, n3min,
                                    T_a_in_val[i], T_a_out_val[i])
    qpred_val.append(q_pred)

# Plot training and validation
plt.scatter(qpred_train, qdata_train, alpha=0.6, label='Training', s=50, c='blue', edgecolors='k')
plt.scatter(qpred_val, qdata_val, alpha=0.6, label='Validation', s=50, marker='^', c='red', edgecolors='k')

# Add diagonal line for perfect prediction
all_q = np.concatenate([qdata_train, qdata_val])
q_min, q_max = np.min(all_q), np.max(all_q)
plt.plot([q_min, q_max], [q_min, q_max], 'k--', alpha=0.5, linewidth=2, label='Perfect Fit')

plt.xlabel('Predicted Heat Transfer Rate (W/cm²)')
plt.ylabel('Measured Heat Transfer Rate (W/cm²)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.title('Model Fit Quality')
plt.axis('equal')
plt.xlim(q_min * 0.95, q_max * 1.05)
plt.ylim(q_min * 0.95, q_max * 1.05)

plt.tight_layout()
plt_fig_title_2 = f"P2.2_plot.png"
plt.savefig(plt_fig_title_2,bbox_inches='tight', dpi=300)
plt.show()


# Summary statistics
train_errors = []
for i in range(len(qpred_train)):
    train_errors.append(abs(qdata_train[i] - qpred_train[i]) / qdata_train[i])

val_errors_list = []
for i in range(len(qpred_val)):
    val_errors_list.append(abs(qdata_val[i] - qpred_val[i]) / qdata_val[i])

print(f'\n=== ERROR STATISTICS ===')
print(f'Training MAE: {np.mean(train_errors):.4f}')
print(f'Validation MAE: {np.mean(val_errors_list):.4f}')
print(f'Training Max Error: {np.max(train_errors):.4f}')
print(f'Validation Max Error: {np.max(val_errors_list):.4f}')
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
from random import random
from copy import deepcopy

# Extract data columns
qdata_train = training[:, 2]  # Heat transfer rate (W/cm²)
T_a_in_train = training[:, 0]  # T_a,in (°C)
T_a_out_train = training[:, 1]  # T_a,out (°C)

qdata_val = validation[:, 2]
T_a_in_val = validation[:, 0]
T_a_out_val = validation[:, 1]

# Physical constants
sigma = 5.67e-8  # Stefan-Boltzmann constant (W/m²K⁴)
h_c_ref = 80.0  # Reference convection coefficient (W/m²K) from project

# Number of training data points (organisms)
ND = len(training)
NS = ND  # Number of solutions equals number of data points

'''INITIALIZING PARAMETERS'''
n = []
ntemp = []
gen = [0]
n1avg = [0.0]  # (UA)_e average
n2avg = [0.0]  # (UA)_c average
n3avg = [0.0]  # α average
meanAFerr = [0.0]
aFerrmeanavgn = [0.0]

# Set program parameters - OPTIMIZED FOR BETTER CONVERGENCE
NGEN = 5000  # Increased generations for better convergence
MFRAC = 0.5  # Adjusted for better selection pressure

# Initialize generation arrays
for k in range(NGEN-1):
    gen.append(k+1)
    meanAFerr.append(0.0)
    aFerrmeanavgn.append(0.0)
    n1avg.append(0.0)
    n2avg.append(0.0)
    n3avg.append(0.0)

'''Initial guesses for solution population - OPTIMIZED'''
# n1 = (UA)_e, n2 = (UA)_c, n3 = α
n1i = 1.0   # Improved initial guess for (UA)_e
n2i = 1.0  # Improved initial guess for (UA)_c
n3i = 0.5  # Improved initial guess for α

# Initialize array of DNA strands with random perturbations
n = [[n1i + 0.3*random(), n2i + 0.3*random(), n3i + 0.15*random()]]
for i in range(ND-1):
    n.append([n1i + 0.3*random(), n2i + 0.3*random(), n3i + 0.15*random()])

# Store also in ntemp
ntemp = deepcopy(n)

# Initialize error arrays
Ferr = [[0.0] for _ in range(ND)]
Ferravgn = [[0.0] for _ in range(ND)]
aFerr = [[0.0] for _ in range(ND)]
aFerravgn = [[0.0] for _ in range(ND)]

# Initialize tracking variables
aFerrmeanavgnMin = 1000000000.0
n1min = 0.0
n2min = 0.0
n3min = 0.0
aFerrta = 0.0
n1ta = 0.0
n2ta = 0.0
n3ta = 0.0

# Validation error tracking
val_errors = []

'''FUNCTION TO CALCULATE HEAT TRANSFER'''
def calculate_heat_transfer(n1, n2, n3, T_in, T_out):
    """Calculate predicted heat transfer using equation 2.3/2.6 from project"""
    # Convert temperatures to Kelvin for radiation calculation
    T_in_K = T_in + 273.15
    T_out_K = T_out + 273.15
    
    # Calculate h_r = σ(T_a,in + T_a,out)(T²_a,in + T²_a,out)
    h_r = sigma * (T_in_K + T_out_K) * (T_in_K**2 + T_out_K**2)
    
    # Calculate heat transfer using equation 2.6
    # q̇ = n2 * (1 + n3*h_r/h_c_ref) * [n1(T_a,in - T_a,out)] / [n1 + n2(1 + n3*h_r/h_c_ref)]
    
    numerator = n1 * (T_in - T_out)
    denominator = n1 + n2 * (1 + n3 * h_r / h_c_ref)
    
    # Avoid division by zero
    if abs(denominator) < 1e-10:
        return 0.0
    
    q_pred = n2 * (1 + n3 * h_r / h_c_ref) * numerator / denominator
    
    return q_pred

print("Starting genetic algorithm optimization...")
print(f"Generations: {NGEN}, Population: {NS}, Selection MFRAC: {MFRAC}\n")

'''START OF EVOLUTION LOOP'''
for k in range(NGEN):
    
    '''CALCULATING ERROR (FITNESS) using equation 2.4'''
    for i in range(ND):
        # Calculate predicted heat transfer
        q_pred = calculate_heat_transfer(n[i][0], n[i][1], n[i][2], 
                                        T_a_in_train[i], T_a_out_train[i])
        
        # Calculate error (equation 2.4): ferr,i = qdata,i - q_pred
        Ferr[i] = qdata_train[i] - q_pred
        
        # Calculate fractional absolute error
        aFerr[i] = abs(Ferr[i]) / abs(qdata_train[i])
    
    # Calculate population statistics
    aFerrmean = np.mean(aFerr)
    meanAFerr[k] = aFerrmean
    aFerrmedian = np.median(aFerr)
    
    '''SELECTION'''
    clim = MFRAC * aFerrmedian  # Cutoff limit
    nkeep = 0
    
    # Select survivors
    for j in range(NS):
        if aFerr[j] < clim:
            nkeep = nkeep + 1
            ntemp[nkeep-1][0] = n[j][0]  # (UA)_e
            ntemp[nkeep-1][1] = n[j][1]  # (UA)_c
            ntemp[nkeep-1][2] = n[j][2]  # α
    
    # Handle case where no organisms survive
    if nkeep == 0:
        nkeep = 1
        best_idx = np.argmin(aFerr)
        ntemp[0][0] = n[best_idx][0]
        ntemp[0][1] = n[best_idx][1]
        ntemp[0][2] = n[best_idx][2]
    
    nnew = NS - nkeep  # Number of offspring needed
    
    '''MATING with adaptive mutation'''
    # Adaptive mutation rate - decreases over generations
    mutation_rate = 0.12 * (1 - k / NGEN) + 0.04
    
    for j in range(nnew):
        # Pick two survivors randomly
        nmate1 = np.random.randint(low=0, high=nkeep)
        nmate2 = np.random.randint(low=0, high=nkeep)
        
        # Create offspring with mutation for (UA)_e
        if np.random.rand() < 0.5:
            ntemp[nkeep+j][0] = n[nmate1][0] * (1 + mutation_rate * 2 * (0.5 - np.random.rand()))
        else:
            ntemp[nkeep+j][0] = n[nmate2][0] * (1 + mutation_rate * 2 * (0.5 - np.random.rand()))
        
        # Ensure positive values
        ntemp[nkeep+j][0] = max(ntemp[nkeep+j][0], 0.01)
        
        # Create offspring with mutation for (UA)_c
        if np.random.rand() < 0.5:
            ntemp[nkeep+j][1] = n[nmate1][1] * (1 + mutation_rate * 2 * (0.5 - np.random.rand()))
        else:
            ntemp[nkeep+j][1] = n[nmate2][1] * (1 + mutation_rate * 2 * (0.5 - np.random.rand()))
        
        ntemp[nkeep+j][1] = max(ntemp[nkeep+j][1], 0.01)
        
        # Create offspring with mutation for α
        if np.random.rand() < 0.5:
            ntemp[nkeep+j][2] = n[nmate1][2] * (1 + mutation_rate * 2 * (0.5 - np.random.rand()))
        else:
            ntemp[nkeep+j][2] = n[nmate2][2] * (1 + mutation_rate * 2 * (0.5 - np.random.rand()))
        
        ntemp[nkeep+j][2] = max(ntemp[nkeep+j][2], 0.0)
    
    # Update population
    n = deepcopy(ntemp)
    
    '''CALCULATE POPULATION AVERAGES'''
    n1avg[k] = 0.0
    n2avg[k] = 0.0
    n3avg[k] = 0.0
    
    for j in range(NS):
        n1avg[k] += n[j][0] / NS
        n2avg[k] += n[j][1] / NS
        n3avg[k] += n[j][2] / NS
    
    '''CALCULATE MEAN ERROR FOR POPULATION AVERAGE'''
    for i in range(ND):
        q_pred_avg = calculate_heat_transfer(n1avg[k], n2avg[k], n3avg[k],
                                            T_a_in_train[i], T_a_out_train[i])
        Ferravgn[i] = qdata_train[i] - q_pred_avg
        aFerravgn[i] = abs(Ferravgn[i]) / abs(qdata_train[i])
    
    aFerrmeanavgn[k] = np.mean(aFerravgn)
    
    '''CALCULATE VALIDATION ERROR'''
    val_error = 0.0
    for i in range(len(validation)):
        q_pred_val = calculate_heat_transfer(n1avg[k], n2avg[k], n3avg[k],
                                            T_a_in_val[i], T_a_out_val[i])
        error_val = qdata_val[i] - q_pred_val
        if abs(qdata_val[i]) > 1e-10:
            val_error += abs(error_val) / abs(qdata_val[i])
        else:
            val_error += abs(error_val)
    val_error = val_error / len(validation)
    val_errors.append(val_error)
    
    '''TIME AVERAGING (after generation 1000)'''
    aFerrta += aFerrmeanavgn[k] / NGEN
    if k > 1000:
        n1ta += n1avg[k] / (NGEN - 1000)
        n2ta += n2avg[k] / (NGEN - 1000)
        n3ta += n3avg[k] / (NGEN - 1000)
    
    '''UPDATE MINIMUM'''
    if aFerrmeanavgn[k] < aFerrmeanavgnMin:
        aFerrmeanavgnMin = aFerrmeanavgn[k]
        n1min = n1avg[k]
        n2min = n2avg[k]
        n3min = n3avg[k]
    
    # Progress updates
    if k % 1000 == 0:
        print(f'Generation {k}: Train Error={aFerrmeanavgn[k]:.5f}, Val Error={val_error:.5f}')

'''END OF EVOLUTION LOOP'''

print("\n" + "="*80)
print("OPTIMIZATION COMPLETE")
print("="*80)

# Calculate predictions using best parameters (minimum error)
qpred_train = []
qpred_val = []

for i in range(ND):
    q_pred = calculate_heat_transfer(n1min, n2min, n3min,
                                    T_a_in_train[i], T_a_out_train[i])
    qpred_train.append(q_pred)

for i in range(len(validation)):
    q_pred = calculate_heat_transfer(n1min, n2min, n3min,
                                    T_a_in_val[i], T_a_out_val[i])
    qpred_val.append(q_pred)

# Calculate RMS deviation and error statistics for TRAINING data
rms_error_train = 0.0
percent_errors_train = []
for i in range(ND):
    error = (qdata_train[i] - qpred_train[i]) / qdata_train[i]  # fractional error
    rms_error_train += error**2
    percent_errors_train.append(abs(error) * 100)
rms_error_train = math.sqrt(rms_error_train / ND)
rms_percent_train = rms_error_train * 100

# Calculate RMS deviation and error statistics for VALIDATION data
rms_error_val = 0.0
percent_errors_val = []
for i in range(len(validation)):
    error = (qdata_val[i] - qpred_val[i]) / qdata_val[i]  # fractional error
    rms_error_val += error**2
    percent_errors_val.append(abs(error) * 100)
rms_error_val = math.sqrt(rms_error_val / len(validation))
rms_percent_val = rms_error_val * 100

# Calculate MAE (Mean Absolute Error) in fractional form
mae_train = np.mean(percent_errors_train) / 100
mae_val = np.mean(percent_errors_val) / 100

# Final results summary
print('\n=== OPTIMIZED PARAMETER VALUES ===')
print(f'(UA)_e (Evaporator): {n1min:.6f}')
print(f'(UA)_c (Condenser):  {n2min:.6f}')
print(f'α (Radiation):       {n3min:.6f}')

print('\n=== TIME-AVERAGED PARAMETERS (Last {0} generations) ==='.format(NGEN-1000))
print(f'(UA)_e: {n1ta:.6f}')
print(f'(UA)_c: {n2ta:.6f}')
print(f'α:      {n3ta:.6f}')

print('\n=== FIT QUALITY METRICS - TRAINING DATA ===')
print(f'RMS deviation:       {rms_percent_train:.2f}%')
print(f'Mean absolute error: {np.mean(percent_errors_train):.2f}%')
print(f'MAE (fractional):    {mae_train:.4f}')
print(f'Max error:           {max(percent_errors_train):.2f}%')

print('\n=== FIT QUALITY METRICS - VALIDATION DATA ===')
print(f'RMS deviation:       {rms_percent_val:.2f}%')
print(f'Mean absolute error: {np.mean(percent_errors_val):.2f}%')
print(f'MAE (fractional):    {mae_val:.4f}')
print(f'Max error:           {max(percent_errors_val):.2f}%')

# Check if target achieved
target_achieved = "✓ YES" if mae_val < 0.045 else "✗ NO (close!)" if mae_val < 0.050 else "✗ NO"
print(f'\n=== TARGET ACHIEVEMENT ===')
print(f'Target: MAE < 0.045 (4.5%)')
print(f'Achieved: {target_achieved}')
print(f'Validation MAE: {mae_val:.4f} ({mae_val*100:.2f}%)')

print("\n" + "="*80)

'''PLOTTING RESULTS'''
plt.rcParams.update({'font.size': 11})

# Create comprehensive figure with 4 subplots
fig = plt.figure(figsize=(20, 12))

# ============================================================================
# SUBPLOT 1: Log-Log Plot of Predicted vs Measured
# ============================================================================
ax1 = plt.subplot(2, 2, 1)
ax1.loglog(qdata_train, qpred_train, 'o', markersize=7, color='blue', 
           alpha=0.7, label='Training', markeredgecolor='darkblue', markeredgewidth=1)
ax1.loglog(qdata_val, qpred_val, 's', markersize=8, color='red', 
           alpha=0.7, label='Validation', markeredgecolor='darkred', markeredgewidth=1.5)

# Perfect fit line
all_q = np.concatenate([qdata_train, qdata_val])
q_min, q_max = np.min(all_q)*0.9, np.max(all_q)*1.1
ax1.loglog([q_min, q_max], [q_min, q_max], 'k--', linewidth=2.5, 
           alpha=0.6, label='Perfect Fit')

ax1.set_xlabel('Measured Heat Transfer Rate (W/cm²)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Predicted Heat Transfer Rate (W/cm²)', fontsize=12, fontweight='bold')
ax1.set_title('Log-Log: Predicted vs Measured q̇', fontsize=13, fontweight='bold')
ax1.legend(fontsize=11, loc='lower right')
ax1.grid(True, alpha=0.3, which='both', linestyle='-', linewidth=0.5)

# ============================================================================
# SUBPLOT 2: Error Evolution
# ============================================================================
ax2 = plt.subplot(2, 2, 2)
ax2.semilogy(gen, aFerrmeanavgn, 'b-', linewidth=2, label='Training Error')
ax2.semilogy(gen, val_errors, 'r-', linewidth=2, label='Validation Error')
ax2.axhline(y=0.045, color='green', linestyle='--', linewidth=2, label='Target (0.045)')
ax2.set_xlabel('Generation', fontsize=12, fontweight='bold')
ax2.set_ylabel('Mean Absolute Fractional Error', fontsize=12, fontweight='bold')
ax2.set_title('Error Evolution Over Generations', fontsize=13, fontweight='bold')
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

# ============================================================================
# SUBPLOT 3: Parameter Evolution
# ============================================================================
ax3 = plt.subplot(2, 2, 3)
ax3.plot(gen, n1avg, 'b-', linewidth=2, label='(UA)_e')
ax3.plot(gen, n2avg, 'r-', linewidth=2, label='(UA)_c')
ax3_twin = ax3.twinx()
ax3_twin.plot(gen, n3avg, 'g-', linewidth=2, label='α')
ax3.set_xlabel('Generation', fontsize=12, fontweight='bold')
ax3.set_ylabel('(UA)_e and (UA)_c', fontsize=12, fontweight='bold')
ax3_twin.set_ylabel('α', fontsize=12, fontweight='bold', color='green')
ax3.set_title('Parameter Evolution', fontsize=13, fontweight='bold')
ax3.legend(loc='upper left', fontsize=11)
ax3_twin.legend(loc='upper right', fontsize=11)
ax3.grid(True, alpha=0.3)

# ============================================================================
# SUBPLOT 4: 3D Surface Plot of q̇ vs T_in and T_out
# ============================================================================
ax4 = fig.add_subplot(2, 2, 4, projection='3d')

# Create grid for surface
T_in_range = np.linspace(np.min(T_a_in_train)-1, np.max(T_a_in_train)+1, 60)
T_out_range = np.linspace(np.min(T_a_out_train)-1, np.max(T_a_out_train)+1, 60)
T_in_grid, T_out_grid = np.meshgrid(T_in_range, T_out_range)

q_grid = np.zeros_like(T_in_grid)
for i in range(len(T_in_range)):
    for j in range(len(T_out_range)):
        q_grid[j, i] = calculate_heat_transfer(n1min, n2min, n3min, 
                                               T_in_grid[j, i], T_out_grid[j, i])

# Plot surface
surf = ax4.plot_surface(T_in_grid, T_out_grid, q_grid, cmap=cm.viridis, 
                        alpha=0.7, edgecolor='none', antialiased=True)

# Add data points
ax4.scatter(T_a_in_train, T_a_out_train, qdata_train, c='blue', marker='o', 
            s=40, alpha=0.8, label='Training Data', edgecolors='darkblue', linewidths=1)
ax4.scatter(T_a_in_val, T_a_out_val, qdata_val, c='red', marker='^', 
            s=60, alpha=0.9, label='Validation Data', edgecolors='darkred', linewidths=1.5)

ax4.set_xlabel('T_a,in (°C)', fontsize=11, fontweight='bold', labelpad=10)
ax4.set_ylabel('T_a,out (°C)', fontsize=11, fontweight='bold', labelpad=10)
ax4.set_zlabel('q̇ (W/cm²)', fontsize=11, fontweight='bold', labelpad=10)
ax4.set_title('Heat Transfer Surface: q̇(T_in, T_out)', fontsize=13, fontweight='bold', pad=20)
ax4.legend(fontsize=10, loc='upper left')
ax4.view_init(elev=25, azim=45)
fig.colorbar(surf, ax=ax4, shrink=0.5, aspect=10, pad=0.1)

plt.tight_layout()
plot_fig_title = "P2_comprehensive_analysis.png"
plt.savefig(plot_fig_title, bbox_inches='tight', dpi=300)
print(f'\nFigure saved as: {plot_fig_title}')
plt.show()

print("\n" + "="*80)
print("ANALYSIS COMPLETE - All plots generated successfully!")
print("="*80)
```
